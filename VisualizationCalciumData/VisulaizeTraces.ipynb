```python
import seaborn as sns
import pandas as pd
import csv
import numpy as np
from scipy.signal import welch
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import random as rd
from past.utils import old_div
```

```python
df = pd.read_csv('CNMFe_001.csv')

cols = df.loc[0].eq(" accepted")
cols = cols.index[cols]
cols = list(cols)

df_data = df[[" ",*cols]].drop([0])
df_data = df_data.astype(np.double)
df_data.loc[6005:,' '] -= 216
df_data.loc[12010:,' '] -= 246
df_data = df_data.drop([287])
df_data.describe()
```

```python
palette = sns.color_palette('viridis',as_cmap=True).colors

step = int(len(palette)/len(cols))

# UNCOMMENT THIS LINE TO MAKE ALL TRACES BLACK
# palette = ['#000' for i in range(len(palette))]

fig, ax = plt.subplots(len(cols), 1)
fig = plt.gcf()
fig.set_size_inches([20,len(cols)*1.322])

color_indices = [i*step for i in range(len(cols))]
rd.shuffle(color_indices)
#Shuffle colors

#color_indices = [0, 100, 25, 200, 50, 175, 75, 125, 150, 225, 40, 160, 200, 30, 135, 205, 15,185,95,215,35, 145, 70]

for i in range(len(cols)):
    sns.lineplot(data=df_data,x=" ",y=cols[i],ax=ax[i],color=palette[color_indices[i]])
    max_val = df_data[cols[i]].max()
    min_val = df_data[cols[i]].min()
    ax[i].plot([300,300],[min_val,max_val],linestyle='dashed',color='grey')
    ax[i].plot([600,600],[min_val,max_val],linestyle='dashed',color='grey')
    ax[i].spines['right'].set_visible(False)
    ax[i].spines['top'].set_visible(False)
    ax[i].spines['left'].set_visible(False)
    ax[i].spines['bottom'].set_visible(False)
    ax[i].set_xlim([0,900])
    
fig.tight_layout()
fig.savefig("CNMFE_v3.jpg",dpi=500)
```

```python
def get_rectangle(x,y,w,h,c):
    return mpatches.Rectangle((x,y),w,h, 
                        fill = True,
                        alpha=0.3,
                        color = c)
```

```python
palette = sns.color_palette('viridis',as_cmap=True).colors

step = int(len(palette)/len(cols))

# UNCOMMENT THIS LINE TO MAKE ALL TRACES BLACK
# palette = ['#000' for i in range(len(palette))]

fig, ax = plt.subplots(len(cols), 1)
fig = plt.gcf()
fig.set_size_inches([20,len(cols)*1.322])

data_x_locations = [[20,5],[30,5],[43,15],[70,15],[122,5],[140,15],[170,15],[200,5],[280,15],
                    [720,5],[740,5],[781,5],[800,7],[820,6],[829,3]]
data_x_colors = [0,0,1,3,4,1,0,1,1,0,1,0,2,0,1]
data_x_palette = sns.color_palette("bright")

color_indices = [i*step for i in range(len(cols))]
rd.shuffle(color_indices)
#Shuffle colors
# color_indices = [0, 100, 25, 200, 50, 175, 75, 125, 150, 225]

for i in range(len(cols)):
    
    max_val = df_data[cols[i]].max()
    min_val = df_data[cols[i]].min()
    
    for j,pair in enumerate(data_x_locations):
        ax[i].add_patch(get_rectangle(pair[0],min_val,pair[1],abs(min_val)+max_val,data_x_palette[data_x_colors[j]]))
    
    sns.lineplot(data=df_data,x=" ",y=cols[i],ax=ax[i],color=palette[color_indices[i]])
    ax[i].plot([300,300],[min_val,max_val],linestyle='dashed',color='grey')
    ax[i].plot([600,600],[min_val,max_val],linestyle='dashed',color='grey')
    
    ax[i].set_xlim([0,900])
    
fig.tight_layout()
fig.savefig("CNMFE_bars_v3.jpg",dpi=500)
```
