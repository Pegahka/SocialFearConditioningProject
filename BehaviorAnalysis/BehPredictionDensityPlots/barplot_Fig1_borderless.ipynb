```python
import matplotlib.pyplot as plt
from scipy.stats import norm, lognorm
import numpy as np

y_p = [0.0045, 0.0071, 0.1021, 0.12, 0.25, 0.23, 0.1323, 0.12, 0.013, 0.021]
y_g = [0.05, 0.24, 0.33, 0.253, 0.076, 0.051]

x_g = [el for el in range(6)]
x_p = [el for el in range(5,15)]
```

```python
X_p = np.linspace(4, 15, 1000)
X_g = np.linspace(-2, 6, 1000)

p = norm.pdf(X_p, 9.48, 1.56)
g = norm.pdf(X_g, 2.41, 1.14)

def width(i):
    return i-0.425
def width_o(i):
    return i+0.425
#plt.bar(list(map(width,x_p)),y_p,color='#078f4f',alpha=0.6,width=0.94)
#plt.bar(list(map(width_o,x_g)),y_g,color="#680aab",alpha=0.6,width=0.94)
plt.bar(list(map(width,x_p)),y_p,color="#680aab",alpha=0.3,width=0.94)
plt.bar(list(map(width_o,x_g)),y_g,color="#078f4f",alpha=0.3,width=0.94)
plt.plot(X_p, p, '#bad', linewidth=5)
plt.plot(X_g, g, '#7d7', linewidth=5)
plt.plot(X_p, p, 'black', linewidth=1)
plt.plot(X_g, g, 'black', linewidth=1)

#colors_boxplot = ['#078f4f', "#680aab"]


plt.title("Approach-avoidance", fontsize=30,y=1.05)
ax = plt.gca()
ax.set_xlabel("Count",fontsize=26)
ax.set_ylabel('Density', fontsize = 28)

ax.set_yticks([0,0.1,0.2,0.3])
ax.set_yticklabels([" ",0.1,0.2,0.3], fontsize = 20)
ax.set_xticks([el*2 for el in range(8)])
ax.set_xlim([-2,15])
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
plt.xticks(fontsize=20)
plt.tight_layout()
plt.savefig('Approach-avoidance.jpg')
plt.savefig('Approach-avoidance.eps', format='eps')
```
