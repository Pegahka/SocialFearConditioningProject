```python
import matplotlib.pyplot as plt
from scipy.stats import norm, lognorm
import numpy as np

y_p = [0.0025, 0.0191, 0.0701, 0.15, 0.27, 0.2, 0.1423, 0.1, 0.024, 0.022]
y_o = [0.03, 0.045, 0.05, 0.1, 0.17, 0.2,0.153, 0.125, 0.076, 0.051]

x_o = [el for el in range(5,15)]
x_p = [el for el in range(5,15)]
```

```python
X_p = np.linspace(4, 15, 1000)
X_o = np.linspace(4, 17, 1000)

p = norm.pdf(X_p, 8.9, 1.48)
o = norm.pdf(X_o, 10.18, 1.93)

def width(i):
    return i-0.425
def width_o(i):
    return i+0.425
#plt.bar(list(map(width,x_p)),y_p,color='#078f4f',alpha=0.6,width=0.94)
#plt.bar(list(map(width_o,x_g)),y_g,color="#680aab",alpha=0.6,width=0.94)
plt.bar(list(map(width_o,x_o)),y_o,color="#ed9121",alpha=0.6,width=0.94)
plt.bar(list(map(width,x_p)),y_p,color="#680aab",alpha=0.3,width=0.94)

plt.plot(X_p, p, '#bad', linewidth=5)
plt.plot(X_o, o, '#fc6', linewidth=5)
plt.plot(X_p, p, 'black', linewidth=1)
plt.plot(X_o, o, 'black', linewidth=1)

#colors_boxplot = ['#078f4f', "#680aab"]


plt.title("Approach-avoidance", fontsize=30,y=1.05)
ax = plt.gca()
ax.set_xlabel("Count",fontsize=26)
ax.set_ylabel('Density', fontsize = 28)

ax.set_yticks([0,0.1,0.2,0.3])
ax.set_yticklabels([" ",0.1,0.2,0.3], fontsize = 20)
ax.set_xticks([el*2 for el in range(10)])
ax.set_xlim([3,18])
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
plt.xticks(fontsize=20)
plt.tight_layout()
plt.savefig('Approach-avoidance_Fig2.jpg')
plt.savefig('Approach-avoidance_Fig2.eps', format='eps')
```
