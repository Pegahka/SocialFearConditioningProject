```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import csv
```

```python
with open('barplot_v6_data1.csv','r') as file:
    reader = csv.reader(file)
    data = list(reader)
    
x_p = []
y_p = []

x_g = []
y_g = []

x_p_grey = []
y_p_grey = []

x_g_grey = []
y_g_grey = []

for el in data:
    if (el[2] == 'purple'):
        x_p.append(int(el[0]))
        x_p_grey.append(int(el[0]))
        y_p.append(float(el[4]))
        y_p_grey.append(float(el[3]))
    else:
        x_g.append(int(el[0]))
        x_g_grey.append(int(el[0]))
        y_g.append(float(el[4]))
        y_g_grey.append(float(el[3]))

data_p = []
for i in range(len(y_p)):
    d = int(y_p[i]*1000)
    for j in range(d):
        data_p.append(x_p[i])
        
data_p_grey = []
for i in range(len(y_p_grey)):
    d = int(y_p_grey[i]*1000)
    for j in range(d):
        data_p_grey.append(x_p_grey[i])
        
data_g = []
for i in range(len(y_g)):
    d = int(y_g[i]*1000)
    for j in range(d):
        data_g.append(x_g[i])
        
data_g_grey = []
for i in range(len(y_g_grey)):
    d = int(y_g_grey[i]*1000)
    for j in range(d):
        data_g_grey.append(x_g_grey[i])
        
prod_g = sum(y_g)
prod_g_grey = sum(y_g_grey)
prod_p = sum(y_p)
prod_p_grey = sum(y_p_grey)

y_g = [el/prod_g for el in y_g]
y_g_grey = [el/prod_g_grey for el in y_g_grey]
y_p = [el/prod_p for el in y_p]
y_p_grey = [el/prod_p_grey for el in y_p_grey]
```

```python
def width(i):
    return i-0.425
def width_o(i):
    return i+0.425

ax = plt.gca()

sns.kdeplot(data_p,color='#bad',bw_adjust=1.5,ax=ax,lw=5)
sns.kdeplot(data_p_grey,color='#bad',bw_adjust=1.5,ax=ax,lw=5)
sns.kdeplot(data_p,color='white',bw_adjust=1.5,ax=ax,lw=1)
sns.kdeplot(data_p_grey,color='grey',bw_adjust=1.5,ax=ax,lw=1)

plt.bar(list(map(width,x_p)),y_p,edgecolor="#680aab",color="white",alpha=0.3,width=0.94,linewidth=2)
plt.bar(list(map(width,x_p_grey)),y_p_grey,edgecolor="#680aab",color="#333",alpha=0.3,width=0.94,linewidth=2)

plt.title("Approach-avoidance", fontsize=30,y=1.05)

ax.set_xlabel("Count",fontsize=26)
ax.set_ylabel('Density', fontsize = 28)

ax.set_yticks([0,0.1,0.2,0.3])
ax.set_yticklabels([" ",0.1,0.2,0.3], fontsize = 20)
ax.set_xticks([el*2 for el in range(8)])
ax.set_xlim([-2,15])
ax.set_ylim([0,0.4])
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
plt.xticks(fontsize=20)
plt.tight_layout()
plt.savefig('Approach-avoidance_purple_v8.jpg')
plt.savefig('Approach-avoidance_purple_v8.eps', format='eps');
```

```python
ax = plt.gca()

sns.kdeplot(data_g,color='#7d7',bw_adjust=1.5,ax=ax,lw=5)
sns.kdeplot(data_g_grey,color='#7d7',bw_adjust=1.5,ax=ax,lw=5)
sns.kdeplot(data_g,color='white',bw_adjust=1.5,ax=ax,lw=1)
sns.kdeplot(data_g_grey,color='grey',bw_adjust=1.5,ax=ax,lw=1)

plt.bar(list(map(width_o,x_g)),y_g,edgecolor="#078f4f",color='white',alpha=0.3,width=0.94,linewidth=2)
plt.bar(list(map(width,x_g_grey)),y_g_grey,edgecolor="#078f4f",color="#333",alpha=0.3,width=0.94,linewidth=2)

plt.title("Approach-avoidance", fontsize=30,y=1.05)

ax.set_xlabel("Count",fontsize=26)
ax.set_ylabel('Density', fontsize = 28)

ax.set_yticks([0,0.1,0.2,0.3])
ax.set_yticklabels([" ",0.1,0.2,0.3], fontsize = 20)
ax.set_xticks([el*2 for el in range(8)])
ax.set_xlim([-2,15])
ax.set_ylim([0,0.4])
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
plt.xticks(fontsize=20)
plt.tight_layout()
plt.savefig('Approach-avoidance_green_v8.jpg')
plt.savefig('Approach-avoidance_green_v8.eps', format='eps');
```
