```python
import matplotlib.pyplot as plt
import csv

with open('data_polar_1.csv',newline='') as f:
    reader = csv.reader(f)
    data_1 = list(reader)
```

```python
data_1_f = []
data_2_f = []

for el in range(2):
    d = []
    g = []
    for el2 in data_1[el]:
        d.append(float(el2))
    for el2 in data_1[el+2]:
        g.append(float(el2))
    data_1_f.append(d)
    data_2_f.append(g)

yt = data_1_f[1]
data_1_f[1] = [(2/(5 /0.25))+ y+0.176 for y in yt]
```

```python
spine_width = 5
grid_width = 5
line_width = 8

colors = ['#078f4f', "#680aab"] #"#85e3ff", "#FF7F50"

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
ax.set_theta_zero_location("N")
ax.set_theta_direction(-1)
ax.set_ylim([0,1])
ax.set_yticks([0,0.25,0.5,0.75,1])
ax.set_yticklabels(["",5,10,15,""])

plt.rc('xtick', labelsize=42) 
plt.rc('ytick', labelsize=42) 

ax.grid(linewidth=grid_width)
ax.spines['polar'].set_linewidth(spine_width)

ax.plot(data_1_f[0],data_1_f[1],color=colors[0],linewidth=line_width)

ax.plot(data_2_f[0],data_2_f[1],color=colors[1],linewidth=line_width)
ax.tick_params(axis="x", direction="out", pad=45)

# Remove the yticks from ax and then replicate a same ax with labels.
ax.set_yticklabels([])  # remove the xticks labels
ax2 = ax.figure.add_axes(ax.get_position(original=True), polar=True, frameon=False) 

ax2.set_xticklabels([],fontname='Arial')
ax2.set_theta_zero_location("N")
ax2.set_theta_direction(-1)
ax2.set_ylim([0,1])
ax2.set_yticks([0,0.25,0.5,0.75,1])
ax2.set_yticklabels(["",5,10,15,""],fontname='Arial')
ax2.set_rlabel_position(225)

fig.set_size_inches(8, 8)

fig.savefig("Fig1_polar.jpg",dpi=600,bbox_inches='tight')
plt.savefig('Fig1_polar.eps', format='eps',dpi=600)
```
