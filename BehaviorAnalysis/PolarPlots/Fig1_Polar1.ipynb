```python
import matplotlib.pyplot as plt
import csv
import random
with open('data_polar_2.csv',newline='') as f:
    reader = csv.reader(f)
    data_1 = list(reader)
```

```python
data_1_f = []
data_2_f = []

for el in range(2):
    d = []
    g = []
    for el2 in data_1[el]:
        d.append(float(el2))
    for el2 in data_1[el+2]:
        g.append(float(el2))
    data_1_f.append(d)
    data_2_f.append(g)

yt = data_2_f[1]
```

```python
spine_width = 2
grid_width = 2
line_width = 2

colors = ['#078f4f', "#680aab"] #"#85e3ff", "#FF7F50"

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
ax.set_theta_zero_location("N")
ax.set_theta_direction(-1)
ax.set_ylim([0,1])
ax.set_yticks([0,0.25,0.5,0.75,1])
ax.set_yticklabels(["",5,10,15,""])

plt.rc('xtick', labelsize=20) 
plt.rc('ytick', labelsize=16) 

ax.tick_params(axis="x", direction="out", pad=14)

ax.grid(linewidth=grid_width)
ax.plot(data_1_f[0],data_1_f[1],color=colors[0],linewidth=line_width)
ax.spines['polar'].set_linewidth(spine_width)
ax.plot(data_2_f[0],yt,color=colors[1],linewidth=line_width)
fig.savefig("Fig1_polar_1.jpg",dpi=500,bbox_inches='tight')
plt.savefig('Fig1_polar_1.eps', format='eps')
```
