```python
import seaborn as sns
import pandas as pd
import csv
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ExctinctionControl.csv')

print(df)
```

```python
data = []

for el in df.drop("Group",axis=1).drop("Control",axis=1):
    for i,val in enumerate(df[el]):
        data.append([el,val,df.iloc[i].get('Control')])
        
newdf = (pd.DataFrame(data,columns=['Experimental Condition','Left preference','control']))
#print(newdf)
```

```python
sns.stripplot(
    data=newdf,
    x='Experimental Condition',
    y='Left preference',
    hue='control',
    jitter=0,
    dodge=True
).axhline(linewidth=1, color='lightgrey',ls='--')

plt.axvline(x=1.5,linewidth=1, color='red',ls='--')

#axvline(x=1.5,linewidth=1, color='red',ls='--') -- how to add in addition?

ax = sns.boxplot(
    data=newdf,
    x='Experimental Condition',
    y='Left preference',
    hue='control',
    palette="Set3",
    fliersize=0
)

handles, labels = ax.get_legend_handles_labels()
l = plt.legend(handles[0:2], ["Control", "Conditioned"])

#ax.set_xticklabels(['Cups Habituation','Cups LTM','Recall LTM'])
ax.set_xticklabels(['Cups','Cups', 'Cups', 'Recall','Recall 2'])

###START: CONNECT THE DOTS
df_list = df.values.tolist()

control_x = [-0.2,0.8,1.8,2.8,3.8]
conditioned_x = [0.2,1.2,2.2,3.2,4.2]

for el in df_list:
    if('y' in el):
        ax.plot(control_x,el[1:-1],'c--')
    else:
        ax.plot(conditioned_x,el[1:-1],'y--')

###END: CONNECT THE DOTS

ax.text(-0.4,1,'Day 1; 24h post conditioning')
#ax.text(0.6,1.05,'Upper side preference')
ax.text(1.6,1,'Day 2; 48h post conditioning')
ax.set_xlabel("Experimental Condition",fontsize=15)
ax.set_ylabel("Preference ratio",fontsize=15)

plt.title("CS- side preference", fontsize=20,y=1.1)


fig = ax.get_figure()
plt.tight_layout()
fig.tight_layout()
fig.savefig("Exp1_ext.jpg",dpi=500)



```
