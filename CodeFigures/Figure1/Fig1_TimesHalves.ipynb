```python
#PKF
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Fig1_TimesHalves.csv')
print(df)
```

```python
# Set the color palette
colors = ['blue', 'blue', 'red', 'red']
colors = ['#078f4f','#078f4f', "#680aab","#680aab"]

# Create a figure and axes
fig, ax = plt.subplots()


# Plot the first two boxplots with filling
sns.boxplot(x='Group', y='Time', data=df, ax=ax,  width=0.55, linewidth=3, palette=colors,fliersize=0)
for i, artist in enumerate(ax.patches):
    col = artist.get_facecolor()
    artist.set_edgecolor(col)
    artist.set_alpha(1)
    for j in range(i*6,i*6+6):
        line = ax.lines[j]
        line.set_color(col)
        line.set_mfc(col)
        line.set_mec(col)
        line.set_alpha(1)
    if i%2==1:        
        artist.set_facecolor('None')
    else:
        artist.set_facecolor('#aba9a9')


# Overlay scatter points on the first two boxplots
sns.stripplot(x='Group', y='Time', data=df, ax=ax, color='#646664', alpha=0.7, size=6,  jitter=0.09)


# Set the labels and title
ax.set_xlabel('')
ax.set_ylabel('Time (s)', fontsize = 30, fontname='Arial')
ax.set_title('')

# Spine parameters
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)

ax.spines['bottom'].set_linewidth(2)
ax.spines['left'].set_linewidth(2)
ax.spines['left'].set_color('black')

# y-axis ticks
ax.set_yticks([0,100,200,300], [0,100,200,300], fontname='Arial')

# x-axis ticks
ax.set_xticks([])
ax.set_xlim([-0.6,3.5])

ax.tick_params(axis='y', which='major', direction='out',length=12, labelsize=24, width=2)
#set figure dimensions
fig.set_size_inches(4,5.2)

# Save figures
plt.tight_layout()
fig.tight_layout()
fig.savefig("Fig1_TimeHalves.jpg",dpi=600,bbox_inches='tight')
plt.savefig('Fig1_TimeHalves.eps', format='eps', dpi=600)

# Show the plot
plt.show()
```
