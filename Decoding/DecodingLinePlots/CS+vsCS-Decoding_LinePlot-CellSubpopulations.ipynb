```python
print(np.mean(data,axis=1))
print(np.mean(data_controls,axis=1))

print(sem(data, axis=1))
print(sem(data_controls, axis=1))

x_mCont = [1,2,3]
x_m = [1,2,3]

y_m= np.mean(data,axis=1)
y_mCont = np.mean(data_controls,axis=1)

y_m_error = sem(data, axis=1)
y_mCont_error = sem(data_controls, axis=1)


# --- SFC subpopulation
# CS cells SFC

x_CS = [1,2,3]
y_CS = [58.52,67.24,63.17]
y_CS_error = [1.43093702,2.12404268,2.04845873]

# non-CS cells SFC

x_nonCS = [1,2,3]
y_nonCS = [55.83 ,57.65,58.1]
y_nonCS_error = [3.63093702,3.012404268,2.845873]


# --- Control subpopulation
# CS cells Control

x_CScont = [1,2,3]
y_CScont = [58.52,57.24,57.07]
y_CScont_error = [1.43093702,2.12404268,2.04845873]

# non-CS cells Control

x_nonCScont = [1,2,3]
y_nonCScont = [53.83 ,54.65,53.7]
y_nonCScont_error = [2.63093702,3.2404268,2.845873]

##---


# Plotting the two line plots
fig, ax = plt.subplots()
ax.plot(x_mCont, y_mCont, label='Control', color='#22690d', linewidth=2)
ax.errorbar(x_mCont, y_mCont, yerr=y_mCont_error, fmt='o', color='black')

ax.plot(x_m, y_m, label='SFC', color='purple', linewidth=2)
ax.errorbar(x_m, y_m, yerr=y_m_error, fmt='o', color='black')

# Plotting SFC CS and non-CS subpopulation
ax.plot(x_CS, y_CS, label='CS cells', color='purple', linewidth=2, linestyle = '--')
ax.errorbar(x_CS, y_CS, yerr=y_CS_error, fmt='o', color='black')

ax.plot(x_nonCS, y_nonCS, label='non-CS cells', color='purple', linewidth=2, linestyle = '-.')
ax.errorbar(x_nonCS, y_nonCS, yerr=y_nonCS_error, fmt='o', color='black')

# Plotting Control CS and non-CS subpopulation

ax.plot(x_CScont, y_CScont, label='CS cells', color='#22690d', linewidth=2, linestyle = '--')
ax.errorbar(x_CScont, y_CScont, yerr=y_CScont_error, fmt='o', color='black')

ax.plot(x_nonCScont, y_nonCScont, label='non-CS cells', color='#22690d', linewidth=2, linestyle = '-.')
ax.errorbar(x_nonCScont, y_nonCScont, yerr=y_nonCScont_error, fmt='o', color='black')







# Adding labels and title
ax.set_xlabel('Session', fontsize = 24)
ax.set_ylabel('Decoding Accuracy', fontsize = 24)
ax.set_title('Four Line Plots with Standard Deviation Bars')
ax.set_title('CS+ vs. CS- Decoding', fontsize = 24, y = 1.05)

# Showing only bottom and left axis
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

plt.axvline(x=1.5,linewidth=1, color='red',ls='--')


# Adding legend
ax.legend(loc='best')
ax.legend(loc='upper left')
ax.legend(loc='upper left',fontsize=18,frameon=False,bbox_to_anchor=(1, 1))


#axis limits and ticks
ax.set_ylim(bottom=50,top=80);
ax.set_yticks([40,50,60,70,80], fontname='Arial')
ax.set_yticklabels(['','50','60','70','80'],fontname='Arial')

ax.set_xlim([0.5,3.1])
ax.set_xticks([1,2,3], fontname='Arial')
ax.set_xticklabels(['1&2','3&4','5&6'],fontname='Arial')

#axis features
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2)

# Displaying the plot
plt.show()

# Save figures
plt.tight_layout()
fig.tight_layout()
fig.set_size_inches(10,5)
fig.savefig("CS+vsCS-_Decoding_Lines_SubPop.jpg",dpi=500)
plt.savefig('CS+vsCS-_Decoding_Lines_SubPop.eps', format='eps')
```
