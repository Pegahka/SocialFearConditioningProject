```python
import matplotlib.pyplot as plt
import numpy as np

# Data for the first line plot
x1 = [1, 2, 3, 4, 5, 6]
y1 = [67.1, 65.1, 64.8, 65, 62, 65]
y1_error = [2.4, 1.8, 2.6, 2.1, 1.7, 1.2]

# Data for the second line plot
x2 = [1, 2, 3, 4, 5, 6]
y2 = [66.2, 63.7, 67, 63, 64.7, 64.8]
y2_error = [2.1, 1.5, 2.2, 1.8, 1.1, 1.5]

# Data for the third line plot
x3 = [1, 2, 3, 4, 5, 6]
y3 = [65.5, 63.5, 79.1, 74.4, 73.3, 75.7]
y3_error = [1.8, 1.4, 3.7, 3.2, 2, 2.8]

# Data for the fourth line plot
x4 = [1, 2, 3, 4, 5, 6]
y4 = [62.4, 63.2, 81.2, 80.1, 77.4, 82.4]
y4_error = [2, 1.4, 3.4, 2.8, 2.7, 2.5]

# Plotting the four line plots
fig, ax = plt.subplots()
ax.plot(x1, y1, label='Stim 1', color='#4bc227', linewidth=2, linestyle = '--')
ax.errorbar(x1, y1, yerr=y1_error, fmt='o', color='black')
ax.plot(x2, y2, label='Stim 2', color='#22690d', linewidth=2)
ax.errorbar(x2, y2, yerr=y2_error, fmt='o', color='black')
ax.plot(x3, y3, label='CS-', color='#aa76de', linewidth=2, linestyle = '--')
ax.errorbar(x3, y3, yerr=y3_error, fmt='o', color='black')
ax.plot(x4, y4, label='CS+', color='purple', linewidth=2)
ax.errorbar(x4, y4, yerr=y4_error, fmt='o', color='black')

# Adding labels and title
ax.set_xlabel('Session', fontsize = 24)
ax.set_ylabel('Decoding Accuracy', fontsize = 24)
ax.set_title('Stimulus mouse Decoding', fontsize = 24, y = 1.05)
ax.set_title('CS vs. Baseline Decoding', fontsize = 24, y = 1.05)


# Showing only bottom and left axis
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

plt.axvline(x=2.5,linewidth=1, color='red',ls='--')


# Adding legend
ax.legend(loc='best')
ax.legend(loc='lower right',fontsize=15,frameon=False)

#axis limits and ticks
ax.set_ylim(bottom=50,top=85);
ax.set_yticks([40,50,60,70,80], fontname='Arial')
ax.set_yticklabels(['','50','60','70','80'],fontname='Arial')

ax.set_xlim([0.5,6.5])
ax.set_xticks([1,2,3,4,5,6], fontname='Arial')
ax.set_xticklabels(['1','2','3','4','5','6'],fontname='Arial')

#axis features
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2)

# Displaying the plot
plt.show()

# Save figures
plt.tight_layout()
fig.tight_layout()
fig.savefig("CS_Baseline_Decoding.jpg",dpi=500)
plt.savefig('CS_Baseline_Decoding.eps', format='eps')
```
