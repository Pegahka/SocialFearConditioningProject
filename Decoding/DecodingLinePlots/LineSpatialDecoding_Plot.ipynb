```python
#include thin lines for individual animals

# Data for the first line plot
x1 = [1, 2, 3]
y1 = [60.675, 59.725, 58.05]
y1_error = [2.4, 1.8, 2.6]

# Data for the second line plot
x2 = [1, 2, 3]
y2 = [59.3, 72.5, 67.1125]
y2_error = [2.1, 1.5, 2.2]


# Plotting the four line plots
fig, ax = plt.subplots()
ax.plot(x1, y1, label='Controls', color='#22690d', linewidth=2)
ax.errorbar(x1, y1, yerr=y1_error, fmt='o', color='black')
ax.plot(x2, y2, label='SFC', color='purple', linewidth=2)
ax.errorbar(x2, y2, yerr=y2_error, fmt='o', color='black')


# Adding labels and title
ax.set_xlabel('Session', fontsize = 24)
ax.set_ylabel('Decoding Accuracy', fontsize = 24)
ax.set_title('Four Line Plots with Standard Deviation Bars')
ax.set_title('Spatial Decoding', fontsize = 24, y = 1.05)

# Showing only bottom and left axis
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

plt.axvline(x=1.5,linewidth=1, color='red',ls='--')


# Adding legend
ax.legend(loc='best')
ax.legend(loc='upper left')

#axis limits and ticks
ax.set_ylim(bottom=50,top=80);
ax.set_yticks([40,50,60,70,80], fontname='Arial')
ax.set_yticklabels(['','50','60','70','80'],fontname='Arial')

ax.set_xlim([0.5,3.1])
ax.set_xticks([1,2,3], fontname='Arial')
ax.set_xticklabels(['1&2','3&4','5&6'],fontname='Arial')

#axis features
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2)

# Displaying the plot
plt.show()

# Save figures
plt.tight_layout()
fig.tight_layout()
fig.savefig("Spatial_Decoding_Lines.jpg",dpi=500)
plt.savefig('Spatial_Decoding_Lines.eps', format='eps')
```
