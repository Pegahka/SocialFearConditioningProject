```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt

#SFC data 

x_sfc = np.array([64.1,59.1,68.2,65.1,64.3, 65.7, 65.1, 64.7, 54.4, 65.9]) #decoding
y_sfc = np.array([0.68,0.72,0.48,0.73, 0.65, 0.57, 0.77, 0.76, 0.21, -0.1]) #behavior

x_soc = np.array(  [55.3,65.5,57.2,62.1,62.6, 59.7, 63.3, 65.1, 58.1, 60.4]) #decoding
y_soc = np.array( [-0.1,0.251,-0.12,0.146,-0.07, 0.01, 0.14, -0.1, 0.08, 0.07]) #behavior

x = np.subtract(x_sfc, x_soc)
y = np.subtract(y_sfc, y_soc)


print(x)
print(y)

print(np.corrcoef(x,y))
print(np.mean(y))
```

```python
fig, ax = plt.subplots()  # create figure and axes

# plot on ax (Axes) object
ax.plot(x, y, 'o',markersize=13,c='purple')

#obtain m (slope) and b(intercept) of linear regression line
m, b = np.polyfit(x, y, 1)


#add linear regression line to scatterplot 
#ax.plot(x, m*x+b, color='#144e96')
ax.plot(x, m*x+b, color='black', linewidth=2)
#ax.set_ylim([0.475,0.85])
ax.set_ylim([-0.3,0.9])
ax.set_yticks([-0.3,0.3,0.9], fontname='Arial')
ax.set_yticklabels(['','0.3','0.9'],fontname='Arial')

#ax.set_xlim([65,80])
ax.set_xlim([-7,12])
ax.set_xticks([-6,0,6,12], fontname='Arial')
ax.set_xticklabels(['-6','0','6','12'],fontname='Arial')

sns.despine(top=True, right=True, left=False, bottom=False)
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.set_xlabel('Δ Decoding accuracy', fontsize=22)
ax.set_ylabel('Δ Discrimination score', fontsize=22)
#ax.tick_params(axis='both', which='major', )
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2)

plt.tight_layout()
fig.tight_layout()
fig.savefig("FigCorrDiffSpatial.jpg",dpi=500)
plt.savefig('FigCorrDiffSpatial.eps', format='eps')
```
