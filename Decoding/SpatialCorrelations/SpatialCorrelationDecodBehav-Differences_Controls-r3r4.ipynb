```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt

#Control cohort 
x_MockExt = np.array([60.1,61.1,64.2,63.7,60.3, 56.7, 62.1, 61.5, 57.4, 62]) #decoding
y_MockExt = np.array([0.08,-0.02,0.1,0.1,-0.07, 0.14, -0.15, -0.18, 0.08, -0.1]) #behavior

x_soc = np.array([61.4,60.9,62.8,61.7,55.6, 62.1, 65.3, 64.5, 54.2, 57.4]) #decoding
y_soc = np.array([0.02,0.1,0.18,-0.03,0.15, -0.09, 0.21, -0.05, 0.14, -0.12]) #behavior

x = np.subtract(x_MockExt, x_soc)
y = np.subtract(y_MockExt, y_soc)

#x = np.absolute(x)
#y = np.absolute(y)

print(x)
print(y)

print(np.corrcoef(x,y))
print(np.mean(y))
```

```python
fig, ax = plt.subplots()  # create figure and axes

# plot on ax (Axes) object
ax.plot(x, y, 'o',markersize=13,c='#22690d')

#obtain m (slope) and b(intercept) of linear regression line
m, b = np.polyfit(x, y, 1)


#add linear regression line to scatterplot 
#ax.plot(x, m*x+b, color='#144e96')
ax.plot(x, m*x+b, color='black', linewidth=2)
#ax.set_ylim([0.475,0.85])
ax.set_ylim([-0.4,0.4])
ax.set_yticks([-0.4,0,0.4], fontname='Arial')
ax.set_yticklabels(['-0.4','0','0.4'],fontname='Arial')

#ax.set_xlim([65,80])
ax.set_xlim([-6,6])
ax.set_xticks([-3,0,3,6], fontname='Arial')
ax.set_xticklabels(['-3','0','3','6'],fontname='Arial')

sns.despine(top=True, right=True, left=False, bottom=False)
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.set_xlabel('Δ Decoding accuracy', fontsize=22)
ax.set_ylabel('Δ Discrimination score', fontsize=22)
#ax.tick_params(axis='both', which='major', )
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2)

plt.tight_layout()
fig.tight_layout()
fig.savefig("FigCorrDiffSpatial_Controls_r3r4.jpg",dpi=500)
plt.savefig('FigCorrDiffSpatial_Controls_r3r4.eps', format='eps')
```
