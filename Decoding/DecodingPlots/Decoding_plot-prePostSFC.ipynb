```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = [
   
      
    [62.1,62.1,58.2,60.1,54.3, 58.3, 64.7, 65.8],
    [54.4,52.9,57.8,52.1,61.1, 57.1, 60.2, 56.3],
   
]

```

```python
labels_x = ["Pre-exposure","Recall 1","Recall 2"]
labels_x = ["Before SFC","After SFC"]
labels_x = ["",""]


text_boxplot = ['***','***']
text_boxplot = ['','']


fig, ax = plt.subplots()
fig.set_size_inches(6, 6)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)

sns.stripplot(
    data=data,
    jitter=0,
    size=6,
    dodge=False,
    palette=['black','black'],
).axhline(y=50,linewidth=2, color='lightgrey',ls='--')


##COLORING THE BORDERS OF BOXPLOTS
PROPS = {
    'boxprops':{'facecolor':'none', 'edgecolor':'#144e96'},
    'medianprops':{'color':'grey'},
    'whiskerprops':{'color':'grey'},
    'capprops':{'color':'grey'}
}

boxplot = sns.boxplot(
    data=data,
    fliersize=0,
    palette=['white','white'],
    width=0.3,
    linewidth=4.5,
    **PROPS
)
#boxplot.set_title("CS+ vs. CS- decoding",fontsize=30)
boxplot.set_xticklabels(labels_x, fontsize=16)
boxplot.set_ylabel("Decoding accuracy (%)", fontsize=25)



boxplot.set_ylim(bottom=50,top=70);
ax.set_yticks([42,50,60,70], fontname='Arial')
ax.set_yticklabels(['','50','60','70'],fontname='Arial')

ax.tick_params(axis='both', which='major', labelsize=16)
ax.spines['left'].set_color('black')
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)

ax.tick_params(axis='y', which='major', direction='in',length=12, labelsize=24, width=2)

ax.set_xticks(['',''])
ax.set_xticklabels(['',''],fontname="Arial", fontsize=20)

ax.legend().set_visible(False)
ax.set_aspect(0.13)

for i,data_row in enumerate(data):
    max_val = max(data_row) + 2
    boxplot.text(s=text_boxplot[i],x=i,y=max_val,ha='center', va='center')
plt.tight_layout()
fig.tight_layout()
fig.savefig("FigDecodeprePost.jpg",dpi=500)
plt.savefig('FigDecodeprePost.eps', format='eps')
```
