```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt


#create data
x = np.array([75.4,62.9,60.8,72.1,62.1,71.1,65.2,67.3])
y = np.array([0.54,0.52,0.32,0.601,0.41, 0.35, 0.48, 0.6])
print(np.corrcoef(x,y))
print(np.mean(y))
```

```python
fig, ax = plt.subplots()  # create figure and axes

# plot on ax (Axes) object
ax.plot(x, y, 'o',markersize=10,c='black')

#obtain m (slope) and b(intercept) of linear regression line
m, b = np.polyfit(x, y, 1)


#add linear regression line to scatterplot 
ax.plot(x, m*x+b, color='#144e96')
ax.set_ylim([0.25,0.85])
ax.set_yticks([0.35, 0.45,0.55,0.65,0.75,0.85], fontname='Arial')
ax.set_yticklabels(['0.35','0.45','0.55','0.65','0.75','0.85'],fontname='Arial')

ax.set_xlim([60,80])
ax.set_xticks([65,70,75,80], fontname='Arial')
ax.set_xticklabels(['65','70','75','80'],fontname='Arial')

sns.despine(top=True, right=True, left=False, bottom=False)
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.set_xlabel('Decoding accuracy', fontsize=22)
ax.set_ylabel('Discrimination score', fontsize=22)
#ax.tick_params(axis='both', which='major', )
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2)

plt.tight_layout()
fig.tight_layout()
fig.savefig("FigCorr_r3r4.jpg",dpi=500)
plt.savefig('FigCorr_r3r4.eps', format='eps')
```
