```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = [
    [52.3,62.1,54.2,65.1,58.6],
    [66.1,57.1,58.2,61.1,63.3, 64.7, 65.1],
    [57.4,62.9,65.8,62.1,52.6, 66.1, 64.2],
]
```

```python
labels_x = ["Pre-exposure","Recall 1","Recall 2"]
text_boxplot = ['***','***','***']

fig, ax = plt.subplots()
fig.set_size_inches(6, 6)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
sns.stripplot(
    data=data,
    jitter=0,
    size=6,
    dodge=True,
    palette=['black','black','black'],
).axhline(y=50,linewidth=1, color='lightgrey',ls='--')


##COLORING THE BORDERS OF BOXPLOTS
PROPS = {
    'boxprops':{'facecolor':'none', 'edgecolor':'slategrey'},
    'medianprops':{'color':'cornflowerblue'},
    'whiskerprops':{'color':'cornflowerblue'},
    'capprops':{'color':'cornflowerblue'}
}

boxplot = sns.boxplot(
    data=data,
    fliersize=0,
    palette=['white','white','white'],
    width=0.5,
    linewidth=2,
    **PROPS
)
# boxplot.set_aspect(0.04)
boxplot.set_title("Spatial decoding",fontsize=30)
boxplot.set_xticklabels(labels_x, fontsize=18)
boxplot.set_ylabel("Decoding accuracy (%)", fontsize=25)
#boxplot.set_xlabel("CS+ vs. CS- decoding",fontsize=14)
boxplot.set_ylim(bottom=39,top=80);
boxplot.set_yticks(np.arange(45,85,10));
ax.tick_params(axis='both', which='major', labelsize=18)


for i,data_row in enumerate(data):
    max_val = max(data_row) + 5
    boxplot.text(s=text_boxplot[i],x=i,y=max_val,ha='center', va='center')
plt.tight_layout()
fig.tight_layout()
fig.savefig("FigDecodeSpatial.jpg",dpi=500)
plt.savefig('FigDecodeSpatialv2.eps', format='eps')
```
