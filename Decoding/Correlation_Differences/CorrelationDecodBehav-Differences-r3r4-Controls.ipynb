```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt


x_soc = np.array([53.3,60.2,63.2,62.1,68.6, 56.1, 57.8, 64.1, 57.1, 55.8]) #decoding
y_soc = np.array([0.02,0.1,0.18,-0.03,0.15, -0.09, 0.21, -0.05, 0.14, -0.12]) #behavior

x_mockExt = np.array([55.3,60.4,64.7,57.4,65.6, 60.1, 59.8, 65.1,  56.4, 52.7]) #decoding
y_mockExt = np.array([0.08,-0.02,0.1,0.1,-0.07, 0.14, -0.15, -0.18, 0.08, -0.1]) #behavior

x = np.subtract(x_mockExt, x_soc)
y = np.subtract(y_mockExt, y_soc)

#y = np.absolute(y)

print(x)
print(y)
print('mean:')
print(np.mean(x))
print(np.mean(y))

print('Pearson\'s correlation:')
print(np.corrcoef(x,y))


```

```python
fig, ax = plt.subplots()  # create figure and axes

# plot on ax (Axes) object
ax.plot(x, y, 'o',markersize=13,c='#22690d')

#obtain m (slope) and b(intercept) of linear regression line
m, b = np.polyfit(x, y, 1)


#add linear regression line to scatterplot 
#ax.plot(x, m*x+b, color='#144e96')
ax.plot(x, m*x+b, color='black', linewidth=2)
#ax.set_ylim([0.475,0.85])
ax.set_ylim([-0.4,0.4])
ax.set_yticks([-0.4,0,0.4], fontname='Arial')
ax.set_yticklabels(['-0.4','0','0.4'],fontname='Arial')

#ax.set_xlim([65,80])
ax.set_xlim([-5,5])
ax.set_xticks([-5,0,5], fontname='Arial')
ax.set_xticklabels(['','0','5'],fontname='Arial')

sns.despine(top=True, right=True, left=False, bottom=False)
ax.spines['left'].set_linewidth(2.5)
ax.spines['bottom'].set_linewidth(2.5)
ax.set_xlabel('Δ Decoding accuracy', fontsize=24)
ax.set_ylabel('Δ Discrimination score', fontsize=24)
ax.set_xlabel('', fontsize=22)
ax.set_ylabel('', fontsize=22)

#ax.tick_params(axis='both', which='major', )
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2.5)

plt.tight_layout()
fig.tight_layout()
fig.savefig("FigCorrDiff_r3r4_Controls.jpg",dpi=500)
plt.savefig('FigCorrDiff_r3r4_Controls.eps', format='eps')
```
