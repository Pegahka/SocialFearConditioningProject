```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt


#create data - include non-learners


x_ext = np.array([65.2,60.8,74.4,62.1,72.1,62.9,60.1,67.3, 71.1, 64.7]) #decoding
y_ext = np.array( [0.47,0.25,0.37,0.68,0.17, 0.3, 0.54, 0.32, 0.04, -0.05]) #behavior

x_soc = np.array( [52.3,60.2,50.2,62.1,62.6, 56.1, 57.8, 54.7, 58.1, 60.1]) #decoding
y_soc = np.array(  [-0.1,0.251,-0.12,0.146,-0.07, 0.01, 0.14, -0.1, 0.08, 0.07]) #behavior

x = np.subtract(x_ext, x_soc)
y = np.subtract(y_ext, y_soc)

x = np.absolute(x)
y = np.absolute(y)

print(x)
print(y)

print(np.corrcoef(x,y))
print(np.mean(y))
```

```python
fig, ax = plt.subplots()  # create figure and axes

# plot on ax (Axes) object
ax.plot(x, y, 'o',markersize=13,c='grey')

#obtain m (slope) and b(intercept) of linear regression line
m, b = np.polyfit(x, y, 1)


#add linear regression line to scatterplot 
#ax.plot(x, m*x+b, color='#144e96')
ax.plot(x, m*x+b, color='black', linewidth=2)
#ax.set_ylim([0.475,0.85])
ax.set_ylim([-0.1,0.8])
ax.set_yticks([-0.1,0.4,0.9], fontname='Arial')
ax.set_yticklabels(['-0.1','0.4','0.9'],fontname='Arial')

#ax.set_xlim([65,80])
ax.set_xlim([0,22])
ax.set_xticks([0,5,10,15,20,25], fontname='Arial')
ax.set_xticklabels(['','5','10','15','20','25'],fontname='Arial')

sns.despine(top=True, right=True, left=False, bottom=False)
ax.spines['left'].set_linewidth(2.5)
ax.spines['bottom'].set_linewidth(2.5)
ax.set_xlabel('Δ Decoding accuracy', fontsize=24)
ax.set_ylabel('|Δ Discrimination score|', fontsize=24)
#ax.tick_params(axis='both', which='major', )
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2.5)

plt.tight_layout()
fig.tight_layout()
fig.savefig("FigCorrDiff_r3r4.jpg",dpi=500)
plt.savefig('FigCorrDiff_r3r4.eps', format='eps')
```
