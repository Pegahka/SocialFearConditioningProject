```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt


#create data
x_soc = np.array([53.3,60.2,63.2,62.1,68.6, 56.1, 57.8, 64.1, 57.1, 55.8])
y_soc = np.array([-0.1,0.37,0.2,0.046,0.01, 0.105, 0.32, -0.05, 0.18, -0.2])

x_mockSFC = np.array([51.1,56.0,63.4,60.1,64.35, 58.4, 55.9, 60.3, 52.7, 58.4])
y_mockSFC = np.array([0,0.20,0.28,-0.026,0.0182, 0.095, 0.182, 0.131, 0.25, 0.05])

x = np.subtract(x_mockSFC, x_soc)
y = np.subtract(y_mockSFC, y_soc)

x = np.absolute(x)
y = np.absolute(y)

print(x)
print(y)

print(np.corrcoef(x,y))
print(np.mean(y))
```

```python
fig, ax = plt.subplots()  # create figure and axes

# plot on ax (Axes) object
ax.plot(x, y, 'o',markersize=10,c='black')

#obtain m (slope) and b(intercept) of linear regression line
m, b = np.polyfit(x, y, 1)


#add linear regression line to scatterplot 
ax.plot(x, m*x+b, color='#144e96')
ax.set_ylim([0,0.3])
ax.set_yticks([0,0.1,0.2,0.3], fontname='Arial')
ax.set_yticklabels(['','0.1','0.2','0.3'],fontname='Arial')

ax.set_xlim([0,5])
ax.set_xticks([0,1,2,3,4,5], fontname='Arial')
ax.set_xticklabels(['0','1','2','3','4','5'],fontname='Arial')

sns.despine(top=True, right=True, left=False, bottom=False)
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.set_xlabel('Δ Decoding accuracy', fontsize=22)
ax.set_ylabel('Δ Discrimination score', fontsize=22)
#ax.tick_params(axis='both', which='major', )
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2)

plt.tight_layout()
fig.tight_layout()
fig.savefig("FigCorr_SocialControls.jpg",dpi=500)
plt.savefig('FigCorr_SocialControls.eps', format='eps')
```
