```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt


#create data - include non-learners
x_sfc = np.array([74.1,77.1,65.2,77.1,70.3, 68.3, 69.7, 78.2, 52.1, 60.3])
y_sfc = np.array([0.68,0.72,0.42,0.73, 0.65, 0.57, 0.7, 0.76, 0.2, -0.3])

x_soc = np.array([52.3,60.2,63.2,62.1,68.6, 56.1, 57.8, 64.1, 58.1, 52.7]) #decoding
y_soc = np.array([-0.1,0.27,0.3,0.146,0.31, 0.205, 0.32, -0.05, 0.1, 0.25]) #behavior

x = np.subtract(x_sfc, x_soc)
y = np.subtract(y_sfc, y_soc)

x = np.absolute(x)
y = np.absolute(y)

print(x)
print(y)

print(np.corrcoef(x,y))
print(np.mean(y))
```

```python
fig, ax = plt.subplots()  # create figure and axes

# plot on ax (Axes) object
ax.plot(x, y, 'o',markersize=10,c='black')

#obtain m (slope) and b(intercept) of linear regression line
m, b = np.polyfit(x, y, 1)


#add linear regression line to scatterplot 
ax.plot(x, m*x+b, color='#144e96')
#ax.set_ylim([0.475,0.85])
ax.set_ylim([0,0.9])
ax.set_yticks([0,0.3,0.6,0.9], fontname='Arial')
ax.set_yticklabels(['','0.3','0.6','0.9'],fontname='Arial')

#ax.set_xlim([65,80])
ax.set_xlim([0,22])
ax.set_xticks([0,5,10,15,20,25], fontname='Arial')
ax.set_xticklabels(['0','5','10','15','20','25'],fontname='Arial')

sns.despine(top=True, right=True, left=False, bottom=False)
ax.spines['left'].set_linewidth(2)
ax.spines['bottom'].set_linewidth(2)
ax.set_xlabel('Δ Decoding accuracy', fontsize=22)
ax.set_ylabel('Δ Discrimination score', fontsize=22)
#ax.tick_params(axis='both', which='major', )
ax.tick_params(axis='both', which='major', direction='in',length=12, labelsize=24, width=2)

plt.tight_layout()
fig.tight_layout()
fig.savefig("FigCorr.jpg",dpi=500)
plt.savefig('FigCorr.eps', format='eps')
```
